```python
import pandas as pd
from azure.ai.ml import MLClient
from azure.identity import DefaultAzureCredential
from sklearn.preprocessing import OneHotEncoder
```

```python
ml_client = MLClient.from_config(credential=DefaultAzureCredential())
data_asset = ml_client.data.get("transaction-train", version="1")

train_df = pd.read_csv(data_asset.path)
train_df.head()
```

```python
enc = OneHotEncoder(sparse_output=False)
```

```python
enco = enc.fit(train_df[['account_type','transaction_direction']])
```

```python
pd.DataFrame(enco.transform(train_df[['account_type','transaction_direction']]),columns=enco.get_feature_names_out())
```
